# Cosine Similarity Analysis for GenAI Embeddings

```python
# Initialize two distinct numerical vectors
first_array = [3, 1, 4, 2]
second_array = [2, 5, 1, 3]

print("First Vector (A):", first_array)
print("Second Vector (B):", second_array)
```

```python
def calculate_dot_product(vec_x, vec_y):
    # Calculate the dot product using a simple iterative loop
    dot_accum = 0
    for idx in range(len(vec_x)):
        dot_accum += vec_x[idx] * vec_y[idx]
    return dot_accum

manual_dot = calculate_dot_product(first_array, second_array)
print("Calculated Manual Dot Product:", manual_dot)
```

```python
import math

def calculate_magnitude(vec):
    # Compute Euclidean norm using mathematical formulation sqrt(sum(x^2))
    sum_of_squares = 0
    for value in vec:
        sum_of_squares += value ** 2
    return math.sqrt(sum_of_squares)

magnitude_first = calculate_magnitude(first_array)
magnitude_second = calculate_magnitude(second_array)

print(f"Magnitude of First Vector: {magnitude_first:.6f}")
print(f"Magnitude of Second Vector: {magnitude_second:.6f}")
```

```python
def calculate_manual_cosine(vec_x, vec_y):
    # Core mathematical formula for cosine similarity
    numerator = calculate_dot_product(vec_x, vec_y)
    denominator = calculate_magnitude(vec_x) * calculate_magnitude(vec_y)
    if denominator == 0:
        return 0.0
    return numerator / denominator

manual_similarity = calculate_manual_cosine(first_array, second_array)
print(f"Manual Cosine Similarity calculation: {manual_similarity:.4f}")
```

```python
import numpy as np

# Perform the calculations using NumPy
np_vec_a = np.array(first_array)
np_vec_b = np.array(second_array)

np_dot = np.dot(np_vec_a, np_vec_b)
np_norm_a = np.linalg.norm(np_vec_a)
np_norm_b = np.linalg.norm(np_vec_b)

np_similarity = np_dot / (np_norm_a * np_norm_b)

print("NumPy Dot Product:", np_dot)
print(f"NumPy Norm of A: {np_norm_a:.6f}")
print(f"NumPy Norm of B: {np_norm_b:.6f}")
print(f"NumPy Cosine Similarity: {np_similarity:.6f}")
```

```python
# Validate and verify correctness across different packages
is_manual_np_close = np.isclose(manual_similarity, np_similarity)
is_np_sk_close = np.isclose(np_similarity, sk_similarity)

if is_manual_np_close and is_np_sk_close:
    print("All three approaches produced consistent cosine similarity values.")
else:
    print("Warning: There is a discrepancy between the calculations!")
```

```python
def calculate_safe_similarity(vec_x, vec_y):
    # Validation: enforce matching dimensions
    if len(vec_x) != len(vec_y):
        raise ValueError("Vectors must have the same dimensions.")

    # Validation: zero-vector check
    mag_x = calculate_magnitude(vec_x)
    mag_y = calculate_magnitude(vec_y)

    if mag_x == 0 or mag_y == 0:
        print("Zero vector detected! Returning 0 to prevent division by zero.")
        return 0.0

    return calculate_dot_product(vec_x, vec_y) / (mag_x * mag_y)

# Testing error handling with unequal dimensions
try:
    calculate_safe_similarity([1, 2], [1, 2, 3])
except ValueError as err:
    print("Dimension verification caught expected error:", err)

# Testing zero-vector safety
zero_sim = calculate_safe_similarity([0, 0, 0], [1, 2, 3])
print("Zero vector similarity result:", zero_sim)
```

```python
# Define distinct testing scenarios
benchmark_cases = {
    "Highly similar vectors": ([1, 2, 3], [1.1, 1.9, 3.1]),
    "Opposite-direction vectors": ([2, 4, 6], [-2, -4, -6]),
    "Orthogonal vectors": ([1, 0, 0], [0, 1, 1]),
    "Different/random vectors": ([4, 1, 9], [7, 3, 2])
}

for scenario, (vector_x, vector_y) in benchmark_cases.items():
    score = calculate_safe_similarity(vector_x, vector_y)
    print(f"{scenario:<30} -> Cosine Similarity: {score:.4f}")
```

```python
# Mock representations of hypothetical documents from a RAG retrieval system
# Document 1: "A document about artificial intelligence and machine learning."
# Document 2: "A document about machine learning and deep learning."
mock_doc_embedding_1 = [0.85, 0.40, 0.15, 0.10, 0.05]
mock_doc_embedding_2 = [0.80, 0.45, 0.05, 0.20, 0.15]

rag_similarity = calculate_safe_similarity(mock_doc_embedding_1, mock_doc_embedding_2)
print(f"Document Embedding Similarity: {rag_similarity:.6f}")

if rag_similarity > 0.75:
    print("Interpretation: The documents are relatively similar based on the cosine similarity value.")
else:
    print("Interpretation: The documents represent different concepts or topics.")
```
