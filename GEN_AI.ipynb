```python
import chromadb

# Create an in-memory ChromaDB client for this assessment
client = chromadb.Client()

# Create a personalized collection
collection = client.create_collection(name="genai_assessment")

print("Collection created:", collection.name)
print("Initial document count:", collection.count())
```

```python
collection.add(
    documents=[
        "Python is widely used for generative AI applications.",
        "ChromaDB stores and searches vector embeddings efficiently.",
        "Large language models can generate human-like text.",
        "Football is a popular sport played around the world.",
        "Retrieval augmented generation combines search with generation."
    ],
    metadatas=[
        {"category": "genai"},
        {"category": "database"},
        {"category": "llm"},
        {"category": "sports"},
        {"category": "rag"}
    ],
    ids=["item1", "item2", "item3", "item4", "item5"]
)

print("Assessment documents added successfully!")
print("Total count:", collection.count())
```

```python
preview = collection.peek(limit=3)

print("Preview of the first three documents:")
for doc_id, doc in zip(preview["ids"], preview["documents"]):
    print(f"{doc_id} -> {doc}")
```

```python
result = collection.get(ids=["item1", "item3"])

print("Selected documents:")
for doc_id, doc, meta in zip(result["ids"], result["documents"], result["metadatas"]):
    print(f"{doc_id} | {meta} | {doc}")
```

```python
results = collection.query(
    query_texts=["How are generative AI and language models used?"],
    n_results=2
)

print("Semantic search results:")
for doc_id, doc, distance in zip(results["ids"][0], results["documents"][0], results["distances"][0]):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")
```

```python
results = collection.query(
    query_texts=["Where can I find information about football?"],
    n_results=3,
    where={"category": "sports"}
)

print("Filtered search results:")
for doc_id, doc, distance in zip(results["ids"][0], results["documents"][0], results["distances"][0]):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")
```

```python
collection.update(
    ids=["item4"],
    documents=["Football and cricket are popular sports in India."],
    metadatas=[{"category": "sports", "updated": True}]
)

updated = collection.get(ids=["item4"])
print("Updated document:", updated["documents"])
print("Updated metadata:", updated["metadatas"])
```

```python
collection.upsert(
    documents=[
        "Retrieval augmented generation uses retrieved context with an LLM.",
        "Artificial intelligence is used in healthcare, education, and business."
    ],
    ids=["item5", "item6"]
)

print("Total count after upsert:", collection.count())
print("Upserted documents:", collection.get(ids=["item5", "item6"])["documents"])
```

```python
collection.delete(ids=["item2"])

print("Total count after deleting item2:", collection.count())
print("Remaining ids:", collection.get()["ids"])
```

```python
collection.delete(where={"category": "llm"})

print("Total count after category deletion:", collection.count())
print("Remaining ids:", collection.get()["ids"])
```

```python
final_state = collection.get()

print("Final document count:", collection.count())
print("Final ChromaDB collection state:")
for doc_id, doc, meta in zip(final_state["ids"], final_state["documents"], final_state["metadatas"]):
    print(f"{doc_id} | {meta} | {doc}")
```
